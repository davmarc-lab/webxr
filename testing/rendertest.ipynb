```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# read data from json
import json

renderData = None
with open('perfTest.json') as f:
    d = json.load(f)
    renderData = d["renderData"]

robotsNum = []
times = []
for elem in renderData:
    robotsNum.append(elem["nrobots"])
    times.append(np.array(elem["times"]).mean())

plt.plot(robotsNum, times, '-ro')
plt.title("Analisi dei tempi di render")
plt.grid()
plt.xlabel("Numero di robot")
plt.ylabel("Tempo di rendering (s)")
plt.show()
```
