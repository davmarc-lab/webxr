```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
x = np.array(
    [
        0.033,
        0.046,
        0.034,
        0.036,
        0.034,
        0.034,
        0.034,
        0.035,
        0.032,
        0.033,
        0.033,
        0.037,
        0.035,
        0.038,
        0.03,
    ]
)
mean = x.mean()
std = np.std(x)

plt.boxplot(x, tick_labels=[""])
plt.title("Tempi di esecuzione per il riconoscimento")
plt.ylabel("Tempo (s)")
plt.xlabel("")
# plt.errorbar(1, mean, yerr=std, fmt="s", color="lightgreen", barsabove=True, capsize=6)
plt.text(
    1.1,
    mean,
    rf"$\mu = {mean:.3f}$" + "\n" + rf"$\sigma = {std:.3f}$",
    verticalalignment="center",
    fontsize=10,
)
plt.show()
```
